# Постановка задачи

Нужно выяснить, какая из языковых моделей будет лучше генерировать текст для подкастов.

## Критерии

Текст нейросетей будет оценен по следующим критериям:
1. **Информативность**. Текст должен быть информативным, иначе подкаст не имеет смысла.
2. **Цепочка мыслей**. Текст должен быть последовательным, а не "скакать" между темами статьи.
3. **Стиль**. Должен быть более разговорным, без лишнего пафоса.

## Примеры

Возьмём пару примеров длинных и коротких статей.

```python
article_small1_path = "./data/article_small1.txt"
article_small2_path = "./data/article_small2.txt"
article_big1_path = "./data/article_small1.txt"
article_big2_path = "./data/article_big2.txt"
```

```python
with open(article_small1_path, "r", encoding="utf-8") as f:
    article_small1 = f.read()

print(article_small1[:100])
```

```python
with open(article_small2_path, "r", encoding="utf-8") as f:
    article_small2 = f.read()

print(article_small2[:100])
```

```python
with open(article_big1_path, "r", encoding="utf-8") as f:
    article_big1 = f.read()

print(article_big1[:100])
```

```python
with open(article_big2_path, "r", encoding="utf-8") as f:
    article_big2 = f.read()

print(article_big2[:100])
```

### Подходы

Коротки статьи будем посылать в модели как есть, а вот длинные будем разбивать на чанки и подавать постепенно.

## Модели

Будем рассматривать следующие модели:
- `gpt-oss:20b`.
- `kimi-k2-thinking`.
- `gemini-3-pro-preview`.

В качестве провайдера будет выступать Ollama.

# Подготовка

Установим все зависимости и инициализируем переменные.

```python
!uv add ollama langchain langchain-text-splitters langchain-ollama
```

```python
gpt_model = "gpt-oss:20b-cloud"
kimi_model = "kimi-k2-thinking:cloud"
gemini_model = "gemini-3-pro-preview"
```

```python
models_params = {
    "temperature": 0.6
}
```

```python
from langchain_core.prompts import ChatPromptTemplate

simple_prompt = ChatPromptTemplate.from_messages([
    ("system", "Ты опытный подкаст-хост. Преобразуй статью в подкаст, сохраняя смысл и информацию. Не выдумывай ничего от себя и не вставляй никаких специальных символов - только текст, без описаний контекста."),
    ("user", "{input}")
])
```

```python
from langchain_ollama import ChatOllama

gpt = ChatOllama(
    model=gpt_model,
    **models_params
)
kimi = ChatOllama(
    model=kimi_model,
    **models_params
)
gemini = ChatOllama(
    model=gemini_model,
    **models_params
)
```
